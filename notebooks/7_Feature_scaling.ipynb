# ***Feature Scaling***

This notebook performs feature scaling using StandardScaler.

Objectives:

- Fit the scaler only on the training data.
- Transform both training and testing features.
- Save the fitted scaler.
- Save the scaled datasets.

```python
import pandas as pd

from pathlib import Path

from sklearn.preprocessing import StandardScaler

import joblib
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
PROJECT_DIR = Path("/content/drive/MyDrive/Underwater Audio Data")

SPLIT_DIR = PROJECT_DIR / "splits"

OUTPUT_DIR = PROJECT_DIR / "scaled"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

MODEL_DIR = PROJECT_DIR / "models"

MODEL_DIR.mkdir(parents=True, exist_ok=True)
```

```python
X_train = pd.read_csv(SPLIT_DIR / "X_train.csv")
X_test = pd.read_csv(SPLIT_DIR / "X_test.csv")

y_train = pd.read_csv(SPLIT_DIR / "y_train.csv")
y_test = pd.read_csv(SPLIT_DIR / "y_test.csv")
```

```python
X_train.head()
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(X_train)
```

```python
X_train_scaled = scaler.transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns
)
```

```python
X_train_scaled.head()
```

```python
print(X_train_scaled.mean().head(10))
```

```python
print(X_train_scaled.std().head(10))
```

```python
joblib.dump(
    scaler,
    MODEL_DIR / "standard_scaler.pkl"
)

print("Scaler saved successfully.")
```

```python
X_train_scaled.to_csv(
    OUTPUT_DIR / "X_train_scaled.csv",
    index=False
)

X_test_scaled.to_csv(
    OUTPUT_DIR / "X_test_scaled.csv",
    index=False
)

y_train.to_csv(
    OUTPUT_DIR / "y_train.csv",
    index=False
)

y_test.to_csv(
    OUTPUT_DIR / "y_test.csv",
    index=False
)

print("Scaled datasets saved successfully.")
```
